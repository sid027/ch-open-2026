The tool design must be good otherwise Claude will agressively default to it's internal web search.

Level 3: interests lived on your laptop only.
Level 4: everyone sees everyone's interests in real time.
The MCP server code barely changed — only the storage backend did.
2. The equaliser
Even if you didn't finish Levels 1–3, you can paste this config
and immediately participate in the shared demo.
3. Auth tokens are real
The Bearer token in the config is what keeps random internet users
from adding topics to the workshop database. This is the simplest
form of MCP server authentication.
4. There are words Claude owns — 'remember', 'forget', 'always', 'never'. If your tool description uses them, you are competing with Claude's own instruction-following system. Design your tool triggers around verbs that are unambiguous actions: add, save, store, track, list, delete.
</pre>

<div class="alert alert-block alert-success">
<pre style="background:transparent; border:none; margin:0; padding:0; color:inherit;">
<b>Key learnings</b> 
1. Token cost
Every extra tool call consumes context window tokens. In production,
a badly designed server costs more money AND is slower.
2. Failure modes
The BAD server sometimes returns conflicting results when Claude
calls search_by_topic AND search_by_country and tries to merge them.
The GOOD server always returns one coherent list.
3. The description IS the interface
The single most impactful line in the GOOD server is not code — it is:
"Applies all relevant filters from the natural language query automatically."
This one sentence tells Claude it never needs to call the tool twice.
4. Less is more
6 narrow tools < 1 well-designed tool.
The discipline required to resist adding "just one more tool" is a
real engineering skill.
</pre>
</div>
